```python
# Objective / Fitness Function
import random
# -----------------------------
def f(x):
    # Example: Sphere function
    return sum(xi ** 2 for xi in x)


# -----------------------------
# Particle Swarm Optimization
# -----------------------------
def pso(num_particles, dimensions, max_iterations,
        c1=2.0, c2=2.0, w=0.7,
        lower_bound=-10, upper_bound=10):

    # 1. Particle initialization
    particles = []

    for _ in range(num_particles):
        position = [
            random.uniform(lower_bound, upper_bound)
            for _ in range(dimensions)
        ]

        velocity = [
            random.uniform(-1, 1)
            for _ in range(dimensions)
        ]

        particles.append({
            "position": position,
            "velocity": velocity,
            "pbest": position.copy(),
            "pbest_value": f(position)
        })

    # Find initial global best
    gbest = min(
        particles,
        key=lambda p: p["pbest_value"]
    )["pbest"].copy()

    gbest_value = f(gbest)

    # 2. Main PSO loop
    for iteration in range(max_iterations):

        # 3. Evaluate particles
        for particle in particles:

            current_value = f(particle["position"])

            # 4. If f(p) is better than f(pbest)
            if current_value < particle["pbest_value"]:
                particle["pbest"] = particle["position"].copy()
                particle["pbest_value"] = current_value

        # 7. gbest = best p in P
        best_particle = min(
            particles,
            key=lambda p: p["pbest_value"]
        )

        if best_particle["pbest_value"] < gbest_value:
            gbest = best_particle["pbest"].copy()
            gbest_value = best_particle["pbest_value"]

        # 8. Update every particle
        for particle in particles:

            position = particle["position"]
            velocity = particle["velocity"]
            pbest = particle["pbest"]

            # 9. Velocity update
            for d in range(dimensions):

                r1 = random.random()
                r2 = random.random()

                velocity[d] = (
                    w * velocity[d]
                    + c1 * r1 * (pbest[d] - position[d])
                    + c2 * r2 * (gbest[d] - position[d])
                )

            # 10. Position update
            for d in range(dimensions):
                position[d] += velocity[d]

                # Keep particle within bounds
                position[d] = max(
                    lower_bound,
                    min(position[d], upper_bound)
                )

        print(
            f"Iteration {iteration + 1}: "
            f"Best = {gbest_value:.6f}, "
            f"Position = {gbest}"
        )

    return gbest, gbest_value


# -----------------------------
# Run PSO
# -----------------------------
best_position, best_value = pso(
    num_particles=30,
    dimensions=2,
    max_iterations=100
)

print("\nFinal Result")
print("Best Position:", best_position)
print("Best Value:", best_value)
```
