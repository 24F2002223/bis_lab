```python
import numpy as np

# Objective function
def fitness(x):
    return np.sum(x ** 2)

# Parameters
wolves = 10
dimensions = 2
iterations = 100

lower, upper = -10, 10

# Initialize wolf positions
X = np.random.uniform(
    lower, upper, (wolves, dimensions)
)

# GWO
for t in range(iterations):

    # Sort wolves based on fitness
    scores = np.array([fitness(x) for x in X])
    order = np.argsort(scores)

    alpha = X[order[0]].copy()
    beta = X[order[1]].copy()
    delta = X[order[2]].copy()

    # Decrease a from 2 to 0
    a = 2 - 2 * t / iterations

    for i in range(wolves):

        positions = []

        for leader in [alpha, beta, delta]:

            r1 = np.random.rand(dimensions)
            r2 = np.random.rand(dimensions)

            A = 2 * a * r1 - a
            C = 2 * r2

            D = np.abs(C * leader - X[i])
            new_position = leader - A * D

            positions.append(new_position)

        # Average the three positions
        X[i] = np.mean(positions, axis=0)

        # Boundary handling
        X[i] = np.clip(X[i], lower, upper)

# Final best solution
best = X[np.argmin([fitness(x) for x in X])]

print("Best Solution:", best)
print("Minimum Fitness:", fitness(best))
```
